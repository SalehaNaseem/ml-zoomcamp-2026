```python
hp = df['horsepower']
median_before = hp.median()
mode_hp = hp.mode().iloc[0]
median_after = hp.fillna(mode_hp).median()
print('Median before:', median_before)
print('Mode:', mode_hp)
print('Median after:', median_after)
```

```python
asia = df.loc[df['origin'] == 'Asia', ['vehicle_weight', 'model_year']].head(7)
X = asia.to_numpy()
print("X (first 7 Asia rows):")
print(X)


XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y

print("\nw:", w)
print("Sum of w:", w.sum())
print("Rounded to 3 decimals:", round(w.sum(), 3))
```
